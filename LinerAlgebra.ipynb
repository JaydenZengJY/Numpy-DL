```python
W = torch.arange(40).reshape(2, 5, 4)
print(W)
print(W.shape) 
print(W.sum())

W_sum_axis0 = W.sum(axis=0)
print(W_sum_axis0) 
print(W_sum_axis0.shape)
```

```python
W_sum_axis1 = W.sum(axis=1) #压缩行，每个矩阵，每行按列对应相加
print(W_sum_axis1)
print(W_sum_axis1.shape)


W_sum_axis2 = W.sum(axis=2) #压缩列，每个矩阵，每列按行对应相加
print(W_sum_axis2)
print(W_sum_axis2.shape)
```

```python
print(W.sum(axis=[0, 1])) #两个维度求和
print(W.sum(axis=[0, 1]).shape) #前两维已经压缩，只剩第三维“4”
```

```python
print(W.float().mean()) #W.float()将整数张量转换为浮点数张量，mean()计算所有元素的平均值
print(W.float().sum() / W.numel())
```

```python
print(W.float().mean(axis=0)) #沿行的方向求平均，即每一列求平均
print(W.float().sum(axis=0) / W.shape[0]) #第0维的长度是2，除以2得到平均值
```

```python
sum_W = W.sum(axis=0, keepdim=True) #保留被压缩的维度，便于后续广播
print(sum_W)
```

```python
W / sum_W #由于上面保留了维度，0维没有被压缩，所以可以直接进行广播运算
```

```python
print(W)
print(W.cumsum(axis=0)) #沿着第0维累积求和
```
